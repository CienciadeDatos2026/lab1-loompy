## Setup inicial
Cargamos las librearias a usar e importamos el dataset. También hacemos un tratamiento básico a los datos.

```python
#Importamos librerias
import math
import random
from typing import List, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

random.seed(0)
np.random.seed(0)
```

```python
import kagglehub
import os

#Traemos el dataset de Properati.
path = kagglehub.dataset_download("alejandroczernikier/properati-argentina-dataset")
archivos = os.listdir(path)
df = pd.read_csv(os.path.join(path, archivos[0]))

#Vamos a limpiar los datos de forma básica (nuestros ejercicios no son sobre limpieza).
COL_PRECIO = "price"
COL_SUPERFICIE = "surface_total"
COL_MONEDA = "currency"
COL_TIPO_PROP = "property_type"

tipo_elegido = "Departamento"

# Filtramos las propiedades que sean departamentos
datos_raw = df[df[COL_TIPO_PROP] == tipo_elegido].copy()
# Limpiar NaN y asignar a datos para el procesamiento posterior
datos = datos_raw.dropna(subset=[COL_PRECIO, COL_SUPERFICIE, COL_MONEDA]).copy()

# Definimos la cotización del dolar para unificar las monedas.
cotizacion_usd_ars = 1499.05

# Convierte los precios de ARS a USD en la columna COL_PRECIO
ars_rows = datos[COL_MONEDA] == "ARS"
if ars_rows.any(): # Solo realiza la conversión si hay filas con ARS
    datos.loc[ars_rows, COL_PRECIO] = datos.loc[ars_rows, COL_PRECIO] / cotizacion_usd_ars

#Descartamos los datos que esten por debajo del percentil 1 o por encima del percentil 99
p_bajo, p_alto = 0.01, 0.99
lim_precio = datos[COL_PRECIO].quantile([p_bajo, p_alto])
lim_superficie = datos[COL_SUPERFICIE].quantile([p_bajo, p_alto])

datos_limpios = datos[
    (datos[COL_PRECIO].between(lim_precio[p_bajo], lim_precio[p_alto])) &
    (datos[COL_SUPERFICIE].between(lim_superficie[p_bajo], lim_superficie[p_alto])) &
    (datos[COL_SUPERFICIE] > 0)
].copy()

# Asignamos los datos limpios a la variable 'datos' para usarla para los ejercicios
datos = datos_limpios

#Mostramos una grafica de prueba
#x = datos[COL_SUPERFICIE].astype(float).tolist()
#y = datos[COL_PRECIO].astype(float).tolist()
#plt.figure(figsize=(7, 4))
#plt.scatter(x, y, alpha=0.3, s=15)
#plt.xlabel(f"{COL_SUPERFICIE} (m²)")
#plt.ylabel(f"{COL_PRECIO} (USD)")
#plt.title("Precio vs. superficie")
#plt.ticklabel_format(style='plain', axis='y')
#plt.show()
```
