## Setup inicial
Cargamos las librearias a usar e importamos el dataset. También hacemos un tratamiento básico a los datos.

```python
#Importamos librerias
import math
import random
from typing import List, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

random.seed(0)
np.random.seed(0)
```

```python
import kagglehub
import os

#Traemos el dataset de Properati.
path = kagglehub.dataset_download("alejandroczernikier/properati-argentina-dataset")
archivos = os.listdir(path)
df = pd.read_csv(os.path.join(path, archivos[0]))

# Columnas del dataset que vamos a usar
COL_PRECIO = "price"
COL_HABITACIONES = "rooms"
COL_SUPERFICIE = "surface_total"
COL_MONEDA = "currency"
COL_TIPO_PROP = "property_type"

tipo_elegido = "Departamento"

# Filtramos las propiedades que sean departamentos.
datos_filtrados_tipo = df[df[COL_TIPO_PROP] == tipo_elegido].copy()

# Definimos una cotización del dolar para unificar las monedas.
cotizacion_usd_ars = 1499.05

# Eliminar NaN para las columnas para habitaciones
datos_rooms_prep = datos_filtrados_tipo.dropna(subset=[COL_PRECIO, COL_HABITACIONES, COL_MONEDA]).copy()

# Convertir los precios de ARS a USD
ars_rows_rooms = datos_rooms_prep[COL_MONEDA] == "ARS"
if ars_rows_rooms.any():
    datos_rooms_prep.loc[ars_rows_rooms, COL_PRECIO] = datos_rooms_prep.loc[ars_rows_rooms, COL_PRECIO] / cotizacion_usd_ars

# Descartar outliers por percentiles
p_bajo, p_alto = 0.01, 0.99
lim_precio_rooms = datos_rooms_prep[COL_PRECIO].quantile([p_bajo, p_alto])
lim_rooms_val = datos_rooms_prep[COL_HABITACIONES].quantile([p_bajo, p_alto])

datos_limpios_habitaciones = datos_rooms_prep[
    (datos_rooms_prep[COL_PRECIO].between(lim_precio_rooms[p_bajo], lim_precio_rooms[p_alto])) &
    (datos_rooms_prep[COL_HABITACIONES].between(lim_rooms_val[p_bajo], lim_rooms_val[p_alto])) &
    (datos_rooms_prep[COL_HABITACIONES] > 0)
].copy()

# La variable 'datos' se usará para los ejercicios 1 y 2.
datos = datos_limpios_habitaciones

# Eliminar NaN para las columnas para superficie
datos_surface_prep = datos_filtrados_tipo.dropna(subset=[COL_PRECIO, COL_SUPERFICIE, COL_MONEDA]).copy()

# Convertir precios de ARS a USD
ars_rows_surface = datos_surface_prep[COL_MONEDA] == "ARS"
if ars_rows_surface.any():
    datos_surface_prep.loc[ars_rows_surface, COL_PRECIO] = datos_surface_prep.loc[ars_rows_surface, COL_PRECIO] / cotizacion_usd_ars

# Descartar outliers por percentiles
lim_precio_surface = datos_surface_prep[COL_PRECIO].quantile([p_bajo, p_alto])
lim_surface_val = datos_surface_prep[COL_SUPERFICIE].quantile([p_bajo, p_alto])

datos_limpios_superficie = datos_surface_prep[
    (datos_surface_prep[COL_PRECIO].between(lim_precio_surface[p_bajo], lim_precio_surface[p_alto])) &
    (datos_surface_prep[COL_SUPERFICIE].between(lim_surface_val[p_bajo], lim_surface_val[p_alto])) &
    (datos_surface_prep[COL_SUPERFICIE] > 0)
].copy()

# La variable 'datos_superficie' se usará para el punto de comparación en Ejercicio 3
datos_superficie = datos_limpios_superficie

print("Dataset importado correctamente")
```
